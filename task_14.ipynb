```python
from IPython.display import Image

import numpy as np
import math
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
alpha = 0.05
sample_size_x = 3
sample_size_y = 2
group_x = np.array([-1.11, -6.10, 2.42])
group_y = np.array([-2.29, -2.91])
variance_x = 2
variance_y = 1
```

```python
Image("pics/task_14_1.jpg")
```

```python
z_statistic = (group_x.mean() - group_y.mean()) / math.sqrt(variance_x / sample_size_x + variance_y / sample_size_y)
print(f"Z-statistic = {z_statistic:.4f}")

# Считаем p-value (правосторонний)
p_value = norm.sf(z_statistic) # Для H₁: a > b
print(f"P-value = {p_value:.4f}")
```

```python
Image("pics/task_14_2.jpg")
```

```python
# Мощность
mean_differences = np.linspace(0, 5, 100)
power = 1 - norm.cdf(norm.ppf(1 - alpha) - mean_differences / math.sqrt(variance_x / sample_size_x + variance_y / sample_size_y))

plt.figure(figsize=(10, 6))
plt.plot(mean_differences, power, color='green', label='Мощность критерия')
plt.axvline(x=0, color='orange', linestyle='--', label='H₀: Δ = 0')
plt.xlabel('Разность средних (Δ = μ₁ - μ₂)')
plt.ylabel('Мощность')
plt.title('Кривая мощности для правостороннего Z-теста (H₁: μ₁ > μ₂)')
plt.legend()
plt.grid(True)
plt.show()
```
